```python
# SESSION 5 & 6: DATA ANALYSIS WITH PANDAS AND MATPLOTLIB

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Sales": [12000, 15000, 18000, 17000, 22000, 25000,
              28000, 27000, 26000, 30000, 35000, 40000],
    "Profit": [2500, 3000, 3500, 3200, 4500, 5000,
               6000, 5800, 5500, 6500, 8000, 9500],
    "Customers": [150, 180, 220, 210, 260, 290,
                  330, 320, 310, 350, 420, 500],
    "Returns": [12, 15, 18, 14, 20, 25,
                22, 21, 19, 24, 30, 35]
}

df = pd.DataFrame(data)
print(df.head())
```

```python
# Line chart
plt.figure(figsize=(8, 5))
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()

# Bar chart
plt.figure(figsize=(8, 5))
plt.bar(df["Month"], df["Profit"])
plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.show()

# Scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df["Customers"], df["Sales"])
plt.title("Customers vs Sales")
plt.xlabel("Number of Customers")
plt.ylabel("Sales")
plt.show()

# Histogram
plt.figure(figsize=(8, 5))
plt.hist(df["Sales"], bins=6)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
# Pie chart
quarters = {
    "Q1": sum(df["Sales"][0:3]),
    "Q2": sum(df["Sales"][3:6]),
    "Q3": sum(df["Sales"][6:9]),
    "Q4": sum(df["Sales"][9:12])
}

plt.figure(figsize=(7, 7))
plt.pie(quarters.values(), labels=quarters.keys(), autopct="%1.1f%%")
plt.title("Sales Contribution by Quarter")
plt.show()

# Box plot
plt.figure(figsize=(8, 5))
plt.boxplot(
    [df["Sales"], df["Profit"], df["Returns"]],
    tick_labels=["Sales", "Profit", "Returns"]
)
plt.title("Distribution of Business Metrics")
plt.show()
```

```python
# Gapminder dataset analysis
url = "https://raw.githubusercontent.com/plotly/datasets/master/gapminderDataFiveYear.csv"
gap = pd.read_csv(url)

print(gap.head())
print("Shape:", gap.shape)
print("Columns:", gap.columns.tolist())
print("Total Countries:", gap["country"].nunique())
print("Total Continents:", gap["continent"].nunique())
print("Missing Values:\n", gap.isnull().sum())

print("Average Life Expectancy:", gap["lifeExp"].mean())
print("Average GDP per Capita:", gap["gdpPercap"].mean())

highest_gdp = gap.loc[gap["gdpPercap"].idxmax()]
print("Highest GDP Record:\n", highest_gdp)
```

```python
# Visualizations using Gapminder data

# Bar chart: average life expectancy by continent
bar_data = gap.groupby("continent")["lifeExp"].mean()

plt.figure(figsize=(8, 5))
plt.bar(bar_data.index, bar_data.values)
plt.xlabel("Continent")
plt.ylabel("Average Life Expectancy")
plt.title("Average Life Expectancy vs Continent")
plt.show()

# Line chart: India, China and United States
countries = gap[gap["country"].isin(["India", "China", "United States"])]

plt.figure(figsize=(8, 5))
for country in countries["country"].unique():
    temp = countries[countries["country"] == country]
    plt.plot(temp["year"], temp["lifeExp"], marker="o", label=country)

plt.xlabel("Year")
plt.ylabel("Life Expectancy")
plt.title("Life Expectancy Over Time")
plt.legend()
plt.show()

# Scatter plot: GDP vs life expectancy
plt.figure(figsize=(8, 5))
plt.scatter(gap["gdpPercap"], gap["lifeExp"], alpha=0.6)
plt.xscale("log")
plt.xlabel("GDP per Capita (Log Scale)")
plt.ylabel("Life Expectancy")
plt.title("GDP per Capita vs Life Expectancy")
plt.grid(True)
plt.show()
```
