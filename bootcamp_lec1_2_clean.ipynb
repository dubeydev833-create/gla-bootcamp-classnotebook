```python
# Conditional statements
age = int(input("Enter the age: "))

if age < 13:
    print("Child")
elif age <= 18:
    print("Teenager")
else:
    print("Adult")
```

```python
# Functions
def check_result(marks):
    if marks >= 27:
        return "Pass"
    else:
        return "Fail"

student1 = check_result(35)
student2 = check_result(20)

print("Result 1:", student1)
print("Result 2:", student2)
```

```python
# While loop
i = 1

while i <= 8:
    print("Hello World", i)
    i += 1

print("After the loop, count =", i)
```
