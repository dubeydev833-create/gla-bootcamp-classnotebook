```python
# SESSION 3 & 4: LISTS, DICTIONARIES, COMPLEX LOOPS & NUMPY FOUNDATION

# List
friends = ["Devkinandan Dubey", "Akash", "Krishna", 55, False, True, 98.8]

print(friends[0])

friends[0] = "Devkinandan Dubey"
print(friends[0])

friends[4] = True
print(friends[4])

print(friends[1:5])
```

```python
# Ordered sequence, indexing and loops
subjects = ["Math", "Physics", "Chemistry", "Java", "ML"]

print(subjects[0])
print(subjects[-1])

numbers = [10, -5, 0, 8, -2, 15]

for num in numbers:
    if num > 0:
        print(num, "Positive")
    elif num < 0:
        print(num, "Negative")
    else:
        print(num, "Zero")
```

```python
# List of dictionaries
students = [
    {"name": "Devkinandan Dubey", "study_hours": 5, "attendance": 80, "marks": 85},
    {"name": "Bob", "study_hours": 3, "attendance": 70, "marks": 35}
]

for s in students:
    print("\nName:", s["name"])

    if s["marks"] > 75:
        print("Performance: Strong")
    elif s["marks"] >= 40:
        print("Performance: Need Improvement")
    else:
        print("Performance: Fail")

    if s["attendance"] < 75:
        print("Warning: Low Attendance")

    if s["study_hours"] < 4:
        print("Warning: Need More Study Time")
```

```python
# Taking student data using a loop
students = []

for i in range(5):
    name = input("Enter student name: ")
    study_hours = int(input("Enter study hours: "))
    attendance = int(input("Enter attendance: "))
    marks = int(input("Enter marks: "))

    student = {
        "name": name,
        "study_hours": study_hours,
        "attendance": attendance,
        "marks": marks
    }

    students.append(student)

for s in students:
    print("\nName:", s["name"])

    if s["marks"] > 75:
        print("Performance: Strong")
    elif s["marks"] >= 40:
        print("Performance: Need Improvement")
    else:
        print("Performance: Fail")

    if s["attendance"] < 75:
        print("Warning: Low Attendance")

    if s["study_hours"] < 4:
        print("Warning: Need More Study Time")
```

```python
# Transaction analysis
transactions = [
    {"transaction_id": "TX101", "amount": 6200, "method": "Credit Card"},
    {"transaction_id": "TX102", "amount": 5500, "method": "Gift Card"},
    {"transaction_id": "TX103", "amount": 1200, "method": "Gift Card"},
    {"transaction_id": "TX104", "amount": 8000, "method": "Gift Card"},
    {"transaction_id": "TX105", "amount": 450, "method": "PayPal"}
]

flag_transactions = []

for tx in transactions:
    if tx["amount"] > 5000 and tx["method"] == "Gift Card":
        flag_transactions.append(tx["transaction_id"])

print("Flagged Gift Card Transactions:", flag_transactions)
```
