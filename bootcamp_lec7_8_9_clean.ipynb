```python
# SESSION 7, 8 & 9: LINEAR ALGEBRA, COSINE SIMILARITY AND NLP

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

student_a = np.array([65, 78])
student_b = np.array([37, 55])

print("Student A vector:", student_a)
print("Student B vector:", student_b)

students = pd.DataFrame({
    "name": ["Aarav", "Riya", "Kabir", "Meera", "Devkinandan Dubey", "Sara"],
    "study_hours": [6, 3, 8, 2, 5, 7],
    "attendance": [85, 70, 90, 60, 78, 92],
    "previous_marks": [78, 55, 88, 35, 65, 84],
    "final_marks": [82, 58, 91, 40, 70, 88]
})
```

```python
# Student similarity visualization
plt.figure(figsize=(8, 5))
plt.scatter(students["study_hours"], students["previous_marks"])

for _, row in students.iterrows():
    plt.text(
        row["study_hours"] + 0.1,
        row["previous_marks"] + 0.5,
        row["name"]
    )

plt.xlabel("Study Hours")
plt.ylabel("Previous Marks")
plt.title("Students Similarity Based on Study Hours and Marks")
plt.show()
```

```python
# Cosine similarity
from sklearn.metrics.pairwise import cosine_similarity

feature_columns = ["study_hours", "attendance", "previous_marks"]
student_vectors = students[feature_columns].values

print(student_vectors)

similarity_matrix = cosine_similarity(student_vectors)

similarity_df = pd.DataFrame(
    similarity_matrix,
    index=students["name"],
    columns=students["name"]
)

print(similarity_df)
```

```python
# Find the closest student to Devkinandan Dubey
target_name = "Devkinandan Dubey"
target_scores = similarity_df[target_name].drop(index=target_name)

closest_student = target_scores.idxmax()
highest_score = target_scores.max()

print(f"The student closest to {target_name} is: {closest_student}")
print(f"Similarity score: {highest_score:.4f}")
```

```python
# Tokenization and vocabulary
sentence = "this movie was not good"
tokens = sentence.lower().split()

print(tokens)

vocab = {
    "this": 1,
    "movie": 2,
    "was": 3,
    "not": 4,
    "good": 5
}

token_ids = [vocab[word] for word in tokens]
print(token_ids)
```

```python
# Simple sentiment analysis
positive_words = ["good", "great", "amazing", "excellent", "love"]
negative_words = ["bad", "terrible", "poor", "hate", "boring"]

def simple_sentiment(sentence):
    words = sentence.lower().replace(".", "").split()
    score = 0

    for word in words:
        if word in positive_words:
            score += 1
        elif word in negative_words:
            score -= 1

    if score > 0:
        return "Positive"
    elif score < 0:
        return "Negative"
    return "Neutral"

print(simple_sentiment("The movie was good"))
print(simple_sentiment("The movie was bad"))
print(simple_sentiment("The movie was not good"))
```

```python
# Memory while processing a sentence
sentence = "the movie was not good"
tokens = sentence.split()

memory = ""

for token in tokens:
    memory = memory + " " + token
    print("Current token:", token)
    print("Memory so far:", memory.strip())
```

```python
# Simple hidden state example
sentence = "the movie was not good"
tokens = sentence.split()

hidden_state = {
    "topic": None,
    "negation_active": False,
    "sentiment_word": None
}

for token in tokens:
    if token in ["movie", "food", "product"]:
        hidden_state["topic"] = token

    if token == "not":
        hidden_state["negation_active"] = True

    if token in ["good", "bad", "great", "terrible"]:
        hidden_state["sentiment_word"] = token

    print("Token:", token)
    print("Hidden state:", hidden_state)
    print("-----")
```

```python
# Simple RNN model
import tensorflow as tf
from tensorflow.keras import layers

vocab_size = 1000
embedding_dim = 16

simple_rnn_model = tf.keras.Sequential([
    layers.Input(shape=(8,)),
    layers.Embedding(input_dim=vocab_size, output_dim=embedding_dim),
    layers.SimpleRNN(32),
    layers.Dense(1, activation="sigmoid")
])

simple_rnn_model.summary()
```
